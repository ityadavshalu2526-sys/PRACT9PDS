**1. Import a CSV dataset using Pandas. bold text**

```python
import pandas as pd

data = {
    "Name": ["Asha", "Rahul", "Priya", "Amit", "Asha", "Neha"],
    "Age": [20, 21, None, 22, 20, None],
    "Marks": [85, None, 90, 78, 85, 88],
    "Course": ["BScIT", "BScIT", "BScIT", "BCom", "BScIT", "BCA"]
}

df = pd.DataFrame(data)

df.to_csv("student_data.csv", index=False)

print("CSV file created successfully!")
```

```python
import pandas as pd

df = pd.read_csv("student_data.csv")

print(df)
```

**Q2. Find missing values in the dataset**

```python
print("Missing values:")
print(df.isnull().sum())
```

**Q3. Replace missing numerical values with the mean**

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```

**Q4. Remove duplicate records**

```python
df = df.drop_duplicates()

print("After removing duplicates:")
print(df)
```

**Q5. Display the cleaned dataset**

```python
print("Cleaned Dataset:")
print(df)
```
